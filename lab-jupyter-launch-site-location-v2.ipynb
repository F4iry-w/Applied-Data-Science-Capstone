# TASK 3: Calculate the distances between a launch site to its proximities


Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.


You can calculate the distance between two points on the map based on their `Lat` and `Long` values using the following method:


```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

_TODO:_ Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
# 1. Define the coordinates of the launch site (CCAFS SLC-40)
launch_site_lat = 28.563197
launch_site_lon = -80.576820

# 2. Define the coordinates of the closest coastline (from your hint)
coastline_lat = 28.56367
coastline_lon = -80.57163

# 3. Calculate the distance using the provided function
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# 4. Print the result so you can see it
print("Distance to closest coastline:", distance_coastline, "km")
```

_TODO:_ After obtained its coordinate, create a `folium.Marker` to show the distance


```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
# distance_marker = folium.Marker(
#    coordinate,
#    icon=DivIcon(
#        icon_size=(20,20),
#        icon_anchor=(0,0),
#        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
#        )
#    )

# 1. Define the coordinates for the coastline pin (using the hint coordinates)
coastline_coordinate = [28.56367, -80.57163]

# 2. Format the distance number so it looks clean (e.g., "0.51 KM" instead of "0.5123456 KM")
distance_text = "{:10.2f} KM".format(distance_coastline)

# 3. Create the Marker and customize it with DivIcon to show the text
distance_marker = folium.Marker(
    location=coastline_coordinate,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % distance_text,
    )
)

# 4. Add the new marker to your site_map
site_map.add_child(distance_marker)

# 5. Display the map to see the new marker
site_map
```

_TODO:_ Draw a `PolyLine` between a launch site to the selected coastline point


```python
# Create a `folium.PolyLine` object using the coastline coordinates and launch site coordinate
# lines=folium.PolyLine(locations=coordinates, weight=1)
# 1. Create a list containing the two coordinates you want to connect
coordinates_to_connect = [
    [launch_site_lat, launch_site_lon], # Point A: The rocket pad (CCAFS SLC-40)
    [coastline_lat, coastline_lon]      # Point B: The coastline beach
]

# 2. Create the PolyLine object
lines = folium.PolyLine(
    locations=coordinates_to_connect, 
    weight=2,         # Thickness of the line
    color='blue',     # Color of the line
    opacity=0.6       # Makes the line slightly transparent
)

# 3. Add the PolyLine to your map
site_map.add_child(lines)

# 4. Display the final map!
site_map
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site
```

```python
# 1. Define the coordinates for the closest POIs (Points of Interest)
highway_coord = [28.56365, -80.57083]   # Samuel C. Phillips Pkwy
railway_coord = [28.57206, -80.58528]   # NASA Railroad
city_coord = [28.40167, -80.60394]      # City of Cape Canaveral

# Create a list of these points to loop through
pois = [
    ("Highway", highway_coord),
    ("Railway", railway_coord),
    ("City", city_coord)
]

# 2. Loop through each POI to calculate distance, add a marker, and draw a line
for poi_name, poi_coord in pois:
    
    # Calculate distance
    distance_poi = calculate_distance(launch_site_lat, launch_site_lon, poi_coord[0], poi_coord[1])
    
    # Create and add the Distance Marker (Text label)
    distance_marker = folium.Marker(
        location=poi_coord,
        icon=DivIcon(
            icon_size=(20,20),
            icon_anchor=(0,0),
            html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_poi),
        )
    )
    site_map.add_child(distance_marker)
    
    # Create and add the PolyLine connecting the launch site to the POI
    lines = folium.PolyLine(
        locations=[[launch_site_lat, launch_site_lon], poi_coord],
        weight=2,
        color='blue', # You can change this to 'green' or 'red' to color-code the lines!
        opacity=0.6
    )
    site_map.add_child(lines)

# 3. Display the final map
site_map
```
